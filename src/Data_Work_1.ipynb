```python
#Importação das bibliotecas que são usadas
import pandas as pd
import numpy as np
import math
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from scipy.stats import norm
import pylab
```

```python
#Definição de variáveis para referências complementares
arquivo = 'Planilha 1 - Memória RAM.xlsx'
coluna = 'Memória_RAM_pct'
```

```python
#Leitura dos dados do Excel
df =  pd.read_excel(arquivo)
```

Questão 3

```python
#Cálculo da média
media = df.Memória_RAM_pct.mean()
```

```python
#Cálculo da mediana
mediana = df.Memória_RAM_pct.median()
```

```python
#Cálculo do desvio padrão
dpad = df.Memória_RAM_pct.std()
```

```python
#Fazer item 6.1.1 do documento Explorando a distribuição Normal
```

```python
#Inserir comentário depois
limit_inf1 = round(media-dpad,2)
limit_inf1
```

```python
#Inserir comentário depois
limit_inf2 = round(media-2*dpad,2)
limit_inf2
```

```python
#Inserir comentário depois
limit_inf3 = round(media-3*dpad,2)
limit_inf3
```

```python
#Inserir comentário depois
limit_sup1 = round(media+dpad,2)
limit_sup1
```

```python
#Inserir comentário depois
limit_sup2 = round(media+2*dpad,2)
limit_sup2
```

```python
#Inserir comentário depois
limit_sup3 = round(media+3*dpad,2)
limit_sup3
```

```python
#Cálculo da quantidade de dados no intervalo 1
quantidade_1 = sum(i>=limit_inf1 and i<=limit_sup1 for i in df[coluna].sort_values())
quantidade_1
```

```python
#Cálculo da quantidade de dados no intervalo 2
quantidade_2 = sum(i>=limit_inf2 and i<=limit_sup2 for i in df[coluna].sort_values())
quantidade_2
```

```python
#Cálculo da quantidade de dados no intervalo 3
quantidade_3 = sum(i>=limit_inf3 and i<=limit_sup3 for i in df[coluna].sort_values())
quantidade_3
```

```python
#Porcentagem no intervalo 1
porcentagem_1 = round((quantidade_1/len(df[coluna]))*100,2)
porcentagem_1
```

```python
#Porcentagem no intervalo 2
porcentagem_2 = round((quantidade_2/len(df[coluna]))*100,2)
porcentagem_2
```

```python
#Porcentagem no intervalo 3
porcentagem_3 = round((quantidade_3/len(df[coluna]))*100,2)
porcentagem_3
```

```python
#Cálculo do quartil 1
Q1 = df.Memória_RAM_pct.quantile(q=0.25)
```

```python
#Cálculo do quartil 2
Q2 = df.Memória_RAM_pct.quantile(q=0.50)
```

```python
#Cálculo do quartil 3
Q3 = df.Memória_RAM_pct.quantile(q=0.75)
```

```python
#Verificando simetria
Comp1 = Q2 - Q1
print(Comp1)
```

```python
#Verificando simetria
Comp2 = Q3 - Q2
print(Comp2)
```
